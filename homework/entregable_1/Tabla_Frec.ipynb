```python
# ========= #
# LIBRERÍAS #
# ========= #

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python
# =================== #
# TIEMPOS DE REACCIÓN #
# =================== #

times = pd.Series([0.110, 0.110, 0.126, 0.112, 0.117, 0.113, 0.135, 0.107, 0.122,
0.113, 0.098, 0.122, 0.105, 0.103, 0.119, 0.100, 0.117, 0.113,
0.124, 0.118, 0.132, 0.108, 0.115, 0.120, 0.107, 0.123, 0.109,
0.117, 0.111, 0.112, 0.101, 0.112, 0.111, 0.119, 0.103, 0.100,
0.108, 0.120, 0.099, 0.102, 0.129, 0.115, 0.121, 0.130, 0.134,
0.118, 0.106, 0.128, 0.094, 0.1114])
```

```python
# =========================================== #
# TABLA DE FRECUENCIAS INTERACTIVA Y ESTÉTICA #
# =========================================== #

def tabla_frecuencias_interactiva(
    serie,
    digitos=2,
    metodo="sturges",
    titulo="Tabla de Frecuencias"
):
    """
    Genera una tabla de frecuencias interactiva y estética
    para variables numéricas continuas.

    Parámetros
    ----------
    serie : pandas.Series o array-like
    digitos : int
        Número de decimales.
    metodo : str
        Método para número de clases:
        - 'sturges'
        - 'sqrt'
        - entero manual
    titulo : str
        Título de la tabla.
    """

    # =====================================================
    # LIMPIEZA DE DATOS
    # =====================================================

    serie = pd.Series(serie).dropna()

    n = len(serie)

    # =====================================================
    # NÚMERO DE CLASES
    # =====================================================

    if metodo == "sturges":
        num_clases = int(np.ceil(1 + np.log2(n)))

    elif metodo == "sqrt":
        num_clases = int(np.ceil(np.sqrt(n)))

    elif isinstance(metodo, int):
        num_clases = metodo

    else:
        raise ValueError(
            "metodo debe ser 'sturges', 'sqrt' o un entero"
        )

    # =====================================================
    # HISTOGRAMA
    # =====================================================

    frecuencia, bins = np.histogram(
        serie,
        bins=num_clases
    )

    # =====================================================
    # MEDIDAS
    # =====================================================

    marcas_clase = (bins[:-1] + bins[1:]) / 2

    intervalos = [
        f"[{bins[i]:.{digitos}f}, {bins[i+1]:.{digitos}f})"
        for i in range(len(bins) - 1)
    ]

    frec_abs_acum = np.cumsum(frecuencia)

    frec_rel = frecuencia / frecuencia.sum()

    frec_rel_acum = np.cumsum(frec_rel)

    porcentaje = frec_rel * 100

    porcentaje_acum = frec_rel_acum * 100

    # =====================================================
    # DATAFRAME
    # =====================================================

    tabla = pd.DataFrame({
        "Clase": range(1, num_clases + 1),

        "Intervalo": intervalos,

        "Marca de Clase": np.round(
            marcas_clase,
            digitos
        ),

        "Frecuencia": frecuencia,

        "Frecuencia Acumulada": frec_abs_acum,

        "Frecuencia Relativa": np.round(
            frec_rel,
            digitos
        ),

        "Frecuencia Relativa Acum.": np.round(
            frec_rel_acum,
            digitos
        ),

        "Porcentaje (%)": np.round(
            porcentaje,
            digitos
        ),

        "Porcentaje Acum. (%)": np.round(
            porcentaje_acum,
            digitos
        )
    })

    # =====================================================
    # TABLA INTERACTIVA
    # =====================================================

    fig = go.Figure(
        data=[
            go.Table(

                # -------------------------------
                # ENCABEZADO
                # -------------------------------

                header=dict(
                    values=list(tabla.columns),

                    fill_color="#1F3B73",

                    font=dict(
                        color="white",
                        size=14
                    ),

                    align="center",

                    height=40
                ),

                # -------------------------------
                # CUERPO
                # -------------------------------

                cells=dict(
                    values=[
                        tabla[col]
                        for col in tabla.columns
                    ],

                    fill_color=[
                        [
                            "#F8FAFC"
                            if i % 2 == 0
                            else "#E2E8F0"
                            for i in range(len(tabla))
                        ]
                    ],

                    align="center",

                    font=dict(
                        color="#111827",
                        size=13
                    ),

                    height=32
                )
            )
        ]
    )

    # =====================================================
    # DISEÑO
    # =====================================================

    fig.update_layout(
        title={
            "text": f"<b>{titulo}</b>",
            "x": 0.5,
            "font": {
                "size": 24
            }
        },

        width=1250,
        height=500,

        margin=dict(
            l=20,
            r=20,
            t=60,
            b=20
        )
    )

    # Mostrar tabla
    fig.show()

    # Retornar DataFrame
    return tabla


tabla = tabla_frecuencias_interactiva(
    times,
    digitos=3,
    metodo="sturges",
    titulo="Distribución de Tiempos de Reacción"
)

tabla
```

```python
# =========================================== #
# TABLA DE FRECUENCIAS PARA VARIABLE NUMÉRICA #
# =========================================== #

def tabla_frecuencias_numerica(serie, digitos=2):

    # Eliminar NaN
    serie = pd.Series(serie).dropna()

    # Regla de Sturges
    num_clases = int(np.ceil(1 + np.log2(len(serie))))

    # Histograma
    frecuencia, bins = np.histogram(serie, bins=num_clases)

    # Marcas de clase
    marcas_clase = (bins[:-1] + bins[1:]) / 2

    # Intervalos
    intervalos = [
        f"[{round(bins[i], digitos)}, {round(bins[i+1], digitos)})"
        for i in range(len(bins) - 1)
    ]

    # Frecuencias relativas
    frec_relativa = frecuencia / frecuencia.sum()

    # Tabla
    tabla = pd.DataFrame({
        "Clase": range(1, num_clases + 1),
        "Intervalo": intervalos,
        "Marca de Clase": np.round(marcas_clase, digitos),
        "Frecuencia Absoluta": frecuencia,
        "Frecuencia Relativa": np.round(frec_relativa, digitos)
    })

    return tabla, bins

# HISTOGRAMA INTERACTIVO CON PLOTLY

def histograma_frecuencia_relativa_px(serie):

    serie = pd.Series(serie).dropna()

    # Número de clases (Sturges)
    num_clases = int(np.ceil(1 + np.log2(len(serie))))

    # Histograma interactivo
    fig = px.histogram(
        x=serie,
        nbins=num_clases,
        histnorm='probability',
        opacity=0.85,
        text_auto='.2%'
    )

    # Mejorar diseño
    fig.update_traces(
        marker_line_color='white',
        marker_line_width=1.5,
        hovertemplate=
        '<b>Intervalo</b>: %{x}<br>' +
        '<b>Frecuencia Relativa</b>: %{y:.3f}<extra></extra>'
    )

    fig.update_layout(
        title={
            'text': 'Histograma de Frecuencia Relativa',
            'x': 0.5,
            'font': {'size': 22}
        },
        xaxis_title='Tiempo de Reacción',
        yaxis_title='Frecuencia Relativa',
        bargap=0.05,
        template='plotly_white',
        width=900,
        height=550
    )

    fig.show()


# DIAGRAMA CIRCULAR INTERACTIVO CON PLOTLY
def diagrama_circular_frecuencia_relativa_px(serie):

    tabla, _ = tabla_frecuencias_numerica(serie)

    fig = px.pie(
        tabla,
        values='Frecuencia Relativa',
        names='Intervalo',
        hole=0.35   # donut chart
    )

    fig.update_traces(
        textinfo='percent+label',
        pull=[0.03] * len(tabla),
        hovertemplate=
        '<b>Intervalo</b>: %{label}<br>' +
        '<b>Frecuencia Relativa</b>: %{value:.3f}<extra></extra>'
    )

    fig.update_layout(
        title={
            'text': 'Diagrama Circular de Frecuencia Relativa',
            'x': 0.5,
            'font': {'size': 22}
        },
        template='plotly_white',
        width=850,
        height=650
    )

    fig.show()

# Histograma
histograma_frecuencia_relativa_px(times)

# Diagrama circular
diagrama_circular_frecuencia_relativa_px(times)
```

```python
# ===================================== #
# FUNCIÓN COMPLETA PARA DATOS AGRUPADOS #
# ===================================== #

def resumen_datos_agrupados_interactivo(
    serie,
    digitos=4,
    metodo="sturges",
    titulo="Resumen Estadístico para Datos Agrupados"
):

    # =====================================================
    # LIMPIEZA
    # =====================================================

    serie = pd.Series(serie).dropna()

    n = len(serie)

    # =====================================================
    # NÚMERO DE CLASES
    # =====================================================

    if metodo == "sturges":
        k = int(np.ceil(1 + np.log2(n)))

    elif metodo == "sqrt":
        k = int(np.ceil(np.sqrt(n)))

    elif isinstance(metodo, int):
        k = metodo

    else:
        raise ValueError(
            "metodo debe ser 'sturges', 'sqrt' o entero"
        )

    # =====================================================
    # HISTOGRAMA
    # =====================================================

    frecuencia, bins = np.histogram(
        serie,
        bins=k
    )

    # =====================================================
    # ELEMENTOS BÁSICOS
    # =====================================================

    marcas = (bins[:-1] + bins[1:]) / 2

    amplitud = bins[1] - bins[0]

    fa = frecuencia
    Fa = np.cumsum(fa)

    # =====================================================
    # MEDIA
    # =====================================================

    media = np.sum(marcas * fa) / n

    # =====================================================
    # VARIANZA Y DESVIACIÓN
    # =====================================================

    varianza = np.sum(
        fa * (marcas - media)**2
    ) / (n - 1)

    desviacion = np.sqrt(varianza)

    cv = (desviacion / media) * 100

    # =====================================================
    # MEDIANA
    # =====================================================

    pos_mediana = n / 2

    clase_mediana = np.where(
        Fa >= pos_mediana
    )[0][0]

    Li = bins[clase_mediana]

    F_prev = (
        0 if clase_mediana == 0
        else Fa[clase_mediana - 1]
    )

    f_med = fa[clase_mediana]

    mediana = Li + (
        ((pos_mediana - F_prev) / f_med)
        * amplitud
    )

    # =====================================================
    # MODA
    # =====================================================

    clase_modal = np.argmax(fa)

    Li_modal = bins[clase_modal]

    f1 = fa[clase_modal]

    f0 = (
        0 if clase_modal == 0
        else fa[clase_modal - 1]
    )

    f2 = (
        0 if clase_modal == len(fa)-1
        else fa[clase_modal + 1]
    )

    moda = Li_modal + (
        ((f1 - f0) /
        ((f1 - f0) + (f1 - f2)))
    ) * amplitud

    # =====================================================
    # CUARTILES
    # =====================================================

    def cuartil(p):

        pos = p * n

        clase = np.where(Fa >= pos)[0][0]

        Li = bins[clase]

        F_prev = (
            0 if clase == 0
            else Fa[clase - 1]
        )

        f = fa[clase]

        return Li + (
            ((pos - F_prev) / f)
            * amplitud
        )

    Q1 = cuartil(0.25)
    Q2 = cuartil(0.50)
    Q3 = cuartil(0.75)

    # =====================================================
    # RANGO
    # =====================================================

    rango = np.max(serie) - np.min(serie)

    # =====================================================
    # ASIMETRÍA
    # =====================================================

    m3 = np.sum(
        fa * (marcas - media)**3
    ) / n

    asimetria = m3 / (desviacion**3)

    # =====================================================
    # CURTOSIS
    # =====================================================

    m4 = np.sum(
        fa * (marcas - media)**4
    ) / n

    curtosis = m4 / (desviacion**4)

    exceso_curtosis = curtosis - 3

    # =====================================================
    # DATAFRAME
    # =====================================================

    resumen = pd.DataFrame({

        "Categoría": [

            "Tendencia Central",
            "Tendencia Central",
            "Tendencia Central",

            "Posición",
            "Posición",
            "Posición",

            "Variabilidad",
            "Variabilidad",
            "Variabilidad",
            "Variabilidad",

            "Forma",

            "Curtosis",
            "Curtosis"
        ],

        "Medida": [

            "Media",
            "Mediana",
            "Moda",

            "Q1",
            "Q2",
            "Q3",

            "Rango",
            "Varianza",
            "Desviación Estándar",
            "Coeficiente de Variación (%)",

            "Asimetría de Fisher",

            "Curtosis",
            "Exceso de Curtosis"
        ],

        "Valor": np.round([

            media,
            mediana,
            moda,

            Q1,
            Q2,
            Q3,

            rango,
            varianza,
            desviacion,
            cv,

            asimetria,

            curtosis,
            exceso_curtosis

        ], digitos)

    })

    # =====================================================
    # TABLA INTERACTIVA
    # =====================================================

    colores_categoria = {
        "Tendencia Central": "#DBEAFE",
        "Posición": "#DCFCE7",
        "Variabilidad": "#FEF3C7",
        "Forma": "#FCE7F3",
        "Curtosis": "#EDE9FE"
    }

    colores_filas = [
        colores_categoria[c]
        for c in resumen["Categoría"]
    ]

    fig = go.Figure(

        data=[

            go.Table(

                # -----------------------------------------
                # HEADER
                # -----------------------------------------

                header=dict(

                    values=[
                        "<b>Categoría</b>",
                        "<b>Medida</b>",
                        "<b>Valor</b>"
                    ],

                    fill_color="#1E293B",

                    font=dict(
                        color="white",
                        size=15
                    ),

                    align="center",

                    height=42
                ),

                # -----------------------------------------
                # CELDAS
                # -----------------------------------------

                cells=dict(

                    values=[
                        resumen["Categoría"],
                        resumen["Medida"],
                        resumen["Valor"]
                    ],

                    fill_color=[
                        colores_filas,
                        colores_filas,
                        colores_filas
                    ],

                    align="center",

                    font=dict(
                        color="#111827",
                        size=13
                    ),

                    height=34
                )
            )
        ]
    )

    # =====================================================
    # LAYOUT
    # =====================================================

    fig.update_layout(

        title=dict(
            text=f"<b>{titulo}</b>",
            x=0.5,
            font=dict(size=24)
        ),

        width=950,
        height=620,

        margin=dict(
            l=20,
            r=20,
            t=70,
            b=20
        )
    )

    fig.show()

    return resumen


# =========================================================
# USO
# =====================================================

resumen = resumen_datos_agrupados_interactivo(
    times,
    digitos=4,
    metodo="sturges",
    titulo="Resumen Estadístico de los Tiempos de Reacción"
)

resumen
```

---

### **Interpretación:**

- Asimetría > 0 → cola hacia la derecha.

- Asimetría < 0 → cola hacia la izquierda.

- Exceso de curtosis > 0 → leptocúrtica.

- Exceso de curtosis < 0 → platicúrtica.

- CV alto → mucha dispersión relativa.
