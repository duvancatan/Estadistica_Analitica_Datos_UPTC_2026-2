### Definición del Problema

El conjunto de datos de Colombia Housing & Income Dataset contiene información sobre viviendas e ingresos en Colombia. El objetivo del análisis exploratorio cuantitativo de este conjunto de datos es identificar patrones, tendencias y relaciones entre las variables para comprender mejor las características de las viviendas y los ingresos en Colombia. Esto puede ayudar a responder preguntas como:

- ¿Cuáles son las características más comunes de las viviendas en Colombia?

- ¿Cómo varían los ingresos de los hogares según la ubicación geográfica?

- ¿Existe una relación entre el tamaño de la vivienda y el ingreso del hogar?

- ¿Cómo afecta el acceso a servicios públicos al valor de mercado de las viviendas?

- ¿Cuáles son los factores que influyen en el índice de pobreza en Colombia?

- ¿Cómo varía el acceso a crédito según el nivel educativo del jefe del hogar?

- ¿Cuál es la distribución de la antigüedad de las viviendas en Colombia?

- ¿Cómo afecta el estrato al precio de mercado de la vivienda?

- ¿Cómo afecta el ingreso del hogar al precio de mercado?

````python
# ============================================= #
# Distribuciones de algunas variables numéricas #
# ============================================= #

from plotly.subplots import make_subplots

vars_num = [
    'area_total_m2',
    'habitaciones',
    'ingreso_del_hogar',
    'valor_de_mercado_de_la_vivienda',
    'antiguedad_de_la_vivienda',
    'indice_de_pobreza'
]

fig = make_subplots(
    rows=3,
    cols=2,
    subplot_titles=[
        f'Distribución de {v}'
        for v in vars_num
    ]
)

for i, var in enumerate(vars_num):

    fila = i // 2 + 1
    columna = i % 2 + 1

    fig.add_trace(

        go.Histogram(
            x=data[var],
            nbinsx=50,
            opacity=0.80,
            name=var,
            hovertemplate=
            '<b>Valor</b>: %{x}<br>' +
            '<b>Frecuencia</b>: %{y}<extra></extra>'
        ),

        row=fila,
        col=columna
    )


fig.update_layout(

    template='plotly_white',

    title={
        'text': 'Distribuciones de Variables Numéricas',
        'x': 0.5,
        'font': {'size': 28}
    },

    width=1300,
    height=1000,

    showlegend=False,

    bargap=0.05
)

fig.update_xaxes(title_text='Valor')
fig.update_yaxes(title_text='Frecuencia')

fig.show()
````

````python
# =============================================== #  
# Matriz de correlación entre variables numéricas #
# =============================================== #

corr = (
    data
    .select_dtypes(include=np.number)
    .corr()
)

mask = np.triu(
    np.ones_like(corr, dtype=bool)
)

corr_mask = corr.mask(mask)

fig = px.imshow(
    corr_mask,
    text_auto='.2f',
    aspect='auto',
    color_continuous_midpoint=0
)

fig.update_traces(
    hovertemplate=
    '<b>Variable X</b>: %{x}<br>' +
    '<b>Variable Y</b>: %{y}<br>' +
    '<b>Correlación</b>: %{z:.2f}<extra></extra>'
)

fig.update_layout(
    template='plotly_white',
    title={
        'text': 'Matriz de Correlación - Variables Numéricas',
        'x': 0.5,
        'font': {'size': 26}
    },
    width=1100,
    height=900,
    coloraxis_colorbar=dict(
        title='Correlación'
    )
)

fig.show()
````

````python
# ============================================= #
# Estadísticas de datos filtrandos por estratos #
# ============================================= #

def eda_estrato(estrato):

    df = data[
        data["estrato"] == estrato
    ]

    desc = (
        df
        .describe()
        .round(2)
    )
    desc_reset = desc.reset_index()

    fig_table = go.Figure(
        data=[
            go.Table(
                header=dict(
                    values=[
                        f"<b>{col}</b>"
                        for col in desc_reset.columns
                    ],
                    fill_color="#1E3A8A",
                    font=dict(
                        color="white",
                        size=14
                    ),
                    align="center"
                ),
                cells=dict(
                    values=[
                        desc_reset[col]
                        for col in desc_reset.columns
                    ],
                    fill_color="#F8FAFC",
                    align="center",
                    font=dict(
                        size=13
                    ),
                    height=32
                )
            )
        ]
    )

    fig_table.update_layout(
        template="plotly_white",
        title={
            "text":
            f"Estadísticas Descriptivas - Estrato {estrato}",
            "x": 0.5,
            "font": {"size": 24}
        },
        width=1200,
        height=450
    )
    fig_table.show()

    fig_heat = px.imshow(
        desc,
        text_auto=True,
        aspect='auto'
    )

    fig_heat.update_layout(
        template='plotly_white',
        title={
            'text':
            f'Heatmap Estadístico - Estrato {estrato}',
            'x': 0.5,
            'font': {'size': 24}
        },
        width=1100,
        height=700
    )

    fig_heat.show()

#============================ #
# CAMBIAR ESTRATO MANUALMENTE #
#============================ #

eda_estrato(2)

# eda_estrato(2)
# eda_estrato(3)
# eda_estrato(4)

# ================================= #
#    EJECUTAR DESDE LA TERMINAL     #
# streamlit run src/estrato_eda.py  #
# ================================= #
````

````python
# ========================================= #
# Relación tamaño vivienda vs ingreso hogar #
# ========================================= #

from scipy.stats import pearsonr

# RELACIÓN ÁREA vs INGRESO
corr, p_valor = pearsonr(
    data['area_total_m2'],
    data['ingreso_del_hogar']
)

print(f'Correlación Pearson: {corr:.3f}')
print(f'p-valor: {p_valor:.5f}')

# SCATTER INTERACTIVO
fig = px.scatter(
    data,
    x='area_total_m2',
    y='ingreso_del_hogar',
    trendline='ols',
    opacity=0.65,
    log_x=True,
    log_y=True,
    hover_data=data.columns
)

fig.update_traces(
    marker=dict(size=8),
    hovertemplate=
    '<b>Área m²</b>: %{x}<br>' +
    '<b>Ingreso</b>: %{y:$,.0f}<extra></extra>'
)

fig.update_layout(
    template='plotly_white',
    title={
        'text':
        f'Relación Área Vivienda vs Ingreso<br>'
        f'<sup>r = {corr:.3f} | p-valor = {p_valor:.5f}</sup>',
        'x':0.5,
        'font':{'size':24}
    },
    xaxis_title='Área Total m² (log)',
    yaxis_title='Ingreso del Hogar (log)',
    width=1100,
    height=650
)

fig.show()
````

````python
# ========================================== #
# Acceso servicios públicos vs valor mercado #
# ========================================== #

servicios = [
    'acceso_a_transporte_publico',
    'acceso_a_internet',
    'acceso_a_agua_potable',
    'acceso_a_gas'
]

# TABLA RESUMEN
resultados = []

for serv in servicios:
    valor_si = data[data[serv]=='Yes']['valor_de_mercado_de_la_vivienda'].mean()
    valor_no = data[data[serv]=='No']['valor_de_mercado_de_la_vivienda'].mean()
    diferencia = ((valor_si / valor_no) - 1) * 100
    resultados.append([
        serv,
        round(valor_si,0),
        round(valor_no,0),
        round(diferencia,1)
    ])

df_resultados = pd.DataFrame(
    resultados,
    columns=[
        'Servicio',
        'Valor Promedio Yes',
        'Valor Promedio No',
        'Diferencia %'
    ]
)

# TABLA INTERACTIVA
fig_table = go.Figure(data=[go.Table(
    header=dict(
        values=[
            '<b>Servicio</b>',
            '<b>Yes</b>',
            '<b>No</b>',
            '<b>Diferencia %</b>'
        ],
        fill_color='#1E3A8A',

        font=dict(
            color='white',
            size=14
        ),
        align='center'
    ),

    cells=dict(

        values=[
            df_resultados[col]
            for col in df_resultados.columns
        ],

        fill_color='#F8FAFC',
        align='center',
        font=dict(size=13),
        height=32
    )
)])

fig_table.update_layout(
    template='plotly_white',
    title={
        'text':'Servicios Públicos vs Valor de Mercado',
        'x':0.5,
        'font':{'size':24}
    },
    width=1100,
    height=400
)

fig_table.show()

# REESTRUCTURAR DATOS
df_serv = data[
    servicios + ['valor_de_mercado_de_la_vivienda']
].melt(
    id_vars='valor_de_mercado_de_la_vivienda',
    var_name='Servicio',
    value_name='Acceso'
)

# BOXPLOT INTERACTIVO
fig_box = px.box(
    df_serv,
    x='Servicio',
    y='valor_de_mercado_de_la_vivienda',
    color='Acceso',
    points='outliers',
    log_y=True
)

fig_box.update_traces(
    boxmean='sd',
    hovertemplate=
    '<b>Servicio</b>: %{x}<br>' +
    '<b>Valor Mercado</b>: %{y:$,.0f}<br>' +
    '<b>Acceso</b>: %{fullData.name}<extra></extra>'
)

fig_box.update_layout(
    template='plotly_white',
    title={
        'text':'Valor de Mercado según Acceso a Servicios',
        'x':0.5,
        'font':{'size':24}
    },
    xaxis_title='Servicio',
    yaxis_title='Valor Mercado (log)',
    width=1300,
    height=700
)

fig_box.update_xaxes(tickangle=20)

fig_box.show()
````

````python
# ========================================== #
# Factores que influyen en índice de pobreza #
# ========================================== #

from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.express as px
from scipy.stats import pearsonr
import pandas as pd

# FACTORES ÍNDICE DE POBREZA
vars_pobreza = [
    'estrato',
    'ingreso_del_hogar',
    'nivel_educativo_del_jefe_del_hogar',
    'area_total_m2'
]

print("\nFACTORES ÍNDICE DE POBREZA\n")

for var in vars_pobreza:
    if data[var].dtype == 'object':
        resumen = (
            data
            .groupby(var)['indice_de_pobreza']
            .mean()
            .sort_values()
            .round(2)
        )
        print(f"\n{var}:\n")
        print(resumen)
    else:
        corr, p = pearsonr(
            data[var],
            data['indice_de_pobreza']
        )

        print(
            f"{var}: "
            f"r = {corr:.3f} | "
            f"p-valor = {p:.5f}"
        )

# SUBPLOTS
fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=[
        'Pobreza por Estrato',
        'Pobreza vs Ingreso',
        'Pobreza por Educación',
        'Pobreza vs Área Vivienda'
    ]
)

# BOXPLOT ESTRATO
for estrato in sorted(data['estrato'].dropna().unique()):
    fig.add_trace(
        go.Box(
            y=data[data['estrato']==estrato]['indice_de_pobreza'],
            name=str(estrato),
            boxmean='sd',
            showlegend=False
        ),
        row=1,
        col=1
    )

# SCATTER INGRESO
fig_scatter = px.scatter(
    data,
    x='ingreso_del_hogar',
    y='indice_de_pobreza',
    trendline='ols',
    opacity=0.60
)

for trace in fig_scatter.data:

    fig.add_trace(
        trace,
        row=1,
        col=2
    )

# BOXPLOT EDUCACIÓN
niveles = (
    data['nivel_educativo_del_jefe_del_hogar']
    .dropna()
    .unique()
)

for nivel in niveles:
    fig.add_trace(
        go.Box(
            y=data[
                data['nivel_educativo_del_jefe_del_hogar']==nivel
            ]['indice_de_pobreza'],
            name=str(nivel),
            showlegend=False
        ),
        row=2,
        col=1
    )

# SCATTER ÁREA
fig_area = px.scatter(
    data,
    x='area_total_m2',
    y='indice_de_pobreza',
    trendline='ols',
    opacity=0.60
)

for trace in fig_area.data:

    fig.add_trace(
        trace,
        row=2,
        col=2
    )

fig.update_layout(
    template='plotly_white',
    title={
        'text':'Factores Asociados al Índice de Pobreza',
        'x':0.5,
        'font':{'size':28}
    },
    width=1400,
    height=1000
)

fig.update_xaxes(tickangle=35,row=2,col=1)

fig.show()
````

````python
# ================================= #
# Distribución antigüedad viviendas #
# ================================= #

print("\nDISTRIBUCIÓN ANTIGÜEDAD\n")
print(
    f"Media: "
    f"{data['antiguedad_de_la_vivienda'].mean():.2f}"
)
print(
    f"Mediana: "
    f"{data['antiguedad_de_la_vivienda'].median():.2f}"
)
print("\nPOR ESTRATO:\n")
print(
    data
    .groupby('estrato')['antiguedad_de_la_vivienda']
    .agg(['mean','median'])
    .round(2)
)

# SUBPLOTS
fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=[
        'Distribución Antigüedad Viviendas',
        'Antigüedad por Estrato'
    ]
)

# HISTOGRAMA
fig_hist = px.histogram(
    data,
    x='antiguedad_de_la_vivienda',
    nbins=50,
    marginal='box',
    opacity=0.80
)

for trace in fig_hist.data:

    fig.add_trace(
        trace,
        row=1,
        col=1
    )

# BOXPLOT
for estrato in sorted(data['estrato'].dropna().unique()):

    fig.add_trace(

        go.Box(
            y=data[
                data['estrato']==estrato
            ]['antiguedad_de_la_vivienda'],
            name=str(estrato),
            boxmean='sd',
            showlegend=False,
            hovertemplate=
            '<b>Estrato</b>: %{x}<br>' +
            '<b>Antigüedad</b>: %{y}<extra></extra>'
        ),

        row=1,
        col=2
    )


fig.update_layout(
    template='plotly_white',
    title={
        'text':'Distribución de Antigüedad de las Viviendas',
        'x':0.5,
        'font':{'size':28}
    },
    width=1400,
    height=600
)

fig.update_xaxes(
    title_text='Antigüedad (años)',
    row=1,
    col=1
)

fig.update_yaxes(
    title_text='Frecuencia',
    row=1,
    col=1
)

fig.update_xaxes(
    title_text='Estrato',
    row=1,
    col=2
)

fig.update_yaxes(
    title_text='Antigüedad (años)',
    row=1,
    col=2
)

fig.show()
````
