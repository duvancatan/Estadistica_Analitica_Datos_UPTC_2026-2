### **Conceptos Clave:**

- **Población**: Conjunto completo de individuos, objetos o eventos que comparten una característica común y sobre los cuales se desea obtener información.

- **Parámetro**: Medida numérica que describe una característica de la población, como la media poblacional o la proporción poblacional.
- **Muestra**: Subconjunto representativo de la población que se selecciona para su análisis, con el objetivo de inferir información sobre la población.

- **Estadístico**: Medida numérica que describe una característica de la muestra, como la media muestral o la proporción muestral.

- **Muestreo**: Proceso de seleccionar una muestra representativa de la población, utilizando técnicas como el muestreo aleatorio simple, el muestreo estratificado o el muestreo por conglomerados, entre otros.

---

### **Tipo de Datos:**
- **Datos Categóricos**: Variables que representan categorías o grupos, como género, estado civil o tipo de vivienda.

- **Datos Numéricos**: Variables que representan cantidades o medidas, como ingresos, edad o precio de la vivienda.

- **Datos Ordinales**: Variables que representan categorías con un orden o jerarquía, como nivel educativo (primaria, secundaria, universidad)  o nivel de satisfacción (bajo, medio, alto).

- **Datos de Tiempo**: Variables que representan momentos o períodos de tiempo, como fecha de nacimiento, fecha de compra o fecha de registro.  

---

### **Datos Rectangulares:**

- **Datos Estructurados**: Datos organizados en un formato tabular, con filas y columnas, donde cada columna representa una variable y cada fila representa una observación o registro.

- **Features**: Características o atributos de los datos que se utilizan para realizar el análisis o el modelado.

- **Target**: Variable objetivo o variable dependiente que se desea predecir o explicar a partir de las features.

- **Outcome**: Variable que se desea predecir o explicar en un problema de modelado.

- **Registro**: Cada fila en un conjunto de datos estructurados, que representa una observación individual o una entidad específica.

## **Dependencia Entre Variables** 
---

</div>

### **Definición**

La dependencia entre variables describe cómo el valor de una variable se relaciona con el valor de otra. En particular, al estudiar la dependencia entre una variable **continua** y una **categórica** buscamos entender si los grupos definidos por la variable categórica tienen valores distintos (y en qué medida) en la variable continua.

- **Variable continua**: toma valores numéricos en un rango (ej. ingresos, edad, temperatura).

- **Variable categórica**: toma valores discretos que representan categorías o clases (ej. género, región, tipo de producto).

---

### **¿Por qué estudiar la dependencia entre variables continuas y categóricas?**

- **Comprender patrones**: Saber si la variable continua varía según la categoría (por ejemplo, ingresos por región) ayuda a identificar diferencias relevantes.

- **Selección de variables**: Identificar variables que aportan información predictiva para modelos supervisados.

- **Formular hipótesis**: Permite plantear y validar hipótesis (por ejemplo, "¿los clientes de un segmento gastan más?").

- **Tomar decisiones**: Influye en estrategias de negocio o políticas basadas en segmentación de grupos.

- **Validar suposiciones**: Muchos modelos requieren independencia entre variables o una relación específica.

---

### **Métodos comunes para analizar la dependencia**

- Comparar estadísticas de resumen (media, mediana, desviación) de la variable continua por categoría.

- Visualizaciones: boxplots, violin plots, gráficos de barras con barras de error.

- Pruebas estadísticas: ANOVA, t-test para diferencias de medias; chi-cuadrado si se discretiza la continua; regresión lineal con variables dummy.

## **1. Variables Continuas**

### **Definición:**

La **Correlación** mide el grado de asociación lineal entre dos variables cuantitativas:


$$\rho = \frac{Cov(X,Y)}{\sigma_X \sigma_Y}$$

---

### **Interpretación**

- $ρ ≈ 1 →$ relación positiva fuerte  
- $ρ ≈ -1 →$ relación negativa fuerte  
- $ρ ≈ 0 →$ no hay relación lineal  

---

###  **Importante**

- Correlación $≠$ causalidad  
- Pearson mide relación lineal  
- Spearman mide relación monótona (más general)

```python
# ================== #
# Carga de Librerías #
# ================== #

import numpy as np                  # Computación numérica (arrays, álgebra, operaciones rápidas) 
import pandas as pd                 # Manipulación y análisis de datos (DataFrames)
import seaborn as sns               # Visualización estadística de alto nivel
import matplotlib.pyplot as plt     # Gráficas básicas y personalización
from scipy import stats             # Pruebas estadísticas y distribuciones
import os                           # Interacción con el sistema operativo (archivos, rutas, carpetas)
import plotly.graph_objects as go   # Gráficos Interactivos
import plotly.express as px
from plotly.subplots import make_subplots
from scipy.stats import pearsonr 
from scipy.stats import spearmanr
```

## ⎇ **Relaciones Lineales**

```python
# ====================================== # 
# Geberación de Datos (Tendencia Lineal) #
# ====================================== # 

np.random.seed(42)
x = np.random.normal(0, 1, 100)
y = -2*x + np.random.normal(0, 1, 100)
```

```python
# ===================== #
# Gráfico de Dispersión #
# ===================== #
 
# CORRELACIÓN PEARSON
corr,p_valor = pearsonr(x,y)

print(f"Correlación Pearson: {corr:.4f}")
print(f"p-valor: {p_valor:.6f}")

# SCATTER + LÍNEA DE TENDENCIA
fig = px.scatter(
    x=x,
    y=y,
    trendline='ols',
    opacity=0.70
)

fig.update_traces(
    marker=dict(size=9),
    hovertemplate=
    '<b>X</b>: %{x}<br>' +
    '<b>Y</b>: %{y}<extra></extra>'
)

# LAYOUT
fig.update_layout(
    template='plotly_white',
    title={
        'text':
        'Relación entre X e Y con Línea de Tendencia'
        f'<br><sup>r = {corr:.4f} | p-valor = {p_valor:.6f}</sup>',
        'x':0.5,
        'font':{'size':26}
    },
    xaxis_title='Variable X',
    yaxis_title='Variable Y',
    width=1100,
    height=700
)

fig.show()
```

```python
# Prueba de Pearson #
corr, p_value = stats.pearsonr(x, y)

print(f"Correlación de Pearson: {corr:.4f}")
print(f"p-value: {p_value:.6f}")

# Interpretación automática
alpha = 0.05
if p_value < alpha:
    print("✅ Se rechaza H0: Existe relación lineal")
else:
    print("❌ No se rechaza H0: No hay evidencia de relación lineal")
```

```python
# Prueba de Spearman #
corr, p_value = stats.spearmanr(x, y)

print(f"Correlación de Spearman: {corr:.4f}")
print(f"p-value: {p_value:.6f}")

# Interpretación automática
alpha = 0.05
if p_value < alpha:
    print("✅ Se rechaza H0: Existe relación monótona")
else:
    print("❌ No se rechaza H0: No hay evidencia de relación monótona")
```

## ⎇ **Relaciones No Lineales**

```python
# Generación de los Datos #
np.random.seed(42)

# Más datos (clave)
x = np.linspace(1, 100, 200)

# Relación NO lineal pero monótona (exponencial suave)
y = np.log(x) + np.random.normal(0, 0.1, 200)
```

```python
# Gráfica #

# CORRELACIÓN SPEARMAN
corr,p_valor = spearmanr(x,y)

print(f"Correlación Spearman: {corr:.4f}")
print(f"p-valor: {p_valor:.6f}")

# SCATTER INTERACTIVO
fig = px.scatter(
    x=x,
    y=y,
    opacity=0.70
)

fig.update_traces(
    marker=dict(size=9),
    hovertemplate=
    '<b>X</b>: %{x}<br>' +
    '<b>Y</b>: %{y}<extra></extra>'
)

# LAYOUT
fig.update_layout(
    template='plotly_white',
    title={
        'text':
        'Relación Monótona No Lineal'
        f'<br><sup>Spearman = {corr:.4f} | p-valor = {p_valor:.6f}</sup>',
        'x':0.5,
        'font':{'size':26}
    },
    xaxis_title='Variable X',
    yaxis_title='Variable Y',
    width=1100,
    height=700
)

fig.show()
```

```python
# Función que realiza las pruebas de Pearson y Spearman #

from scipy import stats

def pruebas_correlacion(x, y, alpha=0.05):
    
    # Pearson #
    corr_p, p_p = stats.pearsonr(x, y)
    
    # Spearman #
    corr_s, p_s = stats.spearmanr(x, y)
    
    print(" Resultados")
    print("")
    
    print("🔹 Pearson (Relación Lineal)")
    print(f"Correlación: {corr_p:.4f}")
    print(f"p-value: {p_p:.6f}")
    
    if p_p < alpha:
        print("✅ Se rechaza H0: Existe relación lineal\n")
    else:
        print("❌ No se rechaza H0: No hay evidencia de relación lineal\n")
    
    print("🔹 Spearman (Relación Monótona)")
    print(f"Correlación: {corr_s:.4f}")
    print(f"p-value: {p_s:.6f}")
    
    if p_s < alpha:
        print("✅ Se rechaza H0: Existe Relación Monótona\n")
    else:
        print("❌ No se rechaza H0: No hay evidencia de Relación Monótona\n")
```

```python
# Función que ejecuta las Pruebas Conjuntas #
pruebas_correlacion(x,y)
```

## **Prueba de Kendall**

### **Definición**

La prueba de Kendall (o Kendall’s $Tau$) es una medida no paramétrica que evalúa la fuerza y dirección de la relación monótona entre dos variables continuas u ordinales.

Se basa en comparar pares de observaciones:

🔹 Pares concordantes: cuando el orden relativo entre dos observaciones es consistente en ambas variables

🔹 Pares discordantes: cuando el orden se invierte

La idea clave es:

si la mayoría de los pares mantienen el mismo orden, hay dependencia.


$$\tau = \frac{C-D}{(1/2)n(n-1)}$$

- $C$ : número de pares concordantes
- $D$ : número de pares discordantes
- $n$ : número de observaciones

</div>

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">  

### **Interpretación**

- $\tau$ ≈ 1 → Relación Creciente Fuerte

- $\tau$ ≈ −1 → Relación Decreciente Fuerte

- $\tau$ ≈ 0 → No hay Relación


### **¿Qué prueba de hipótesis realiza?**

$H_0$ : Las variables son Independientes

$H_a$ : Existe Relación Monótona

</div>

```python
import numpy as np
import pandas as pd
from scipy import stats
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Función que ejecuta la prueba de Kendall #
def prueba_kendall(x, y, alpha=0.05):
    
    tau, p_value = stats.kendalltau(x, y)
    
    print("🔹 Prueba de Kendall \n")
    print(f"Tau: {tau:.4f}")
    print(f"p-value: {p_value:.6f}\n")
    
    # Decisión
    if p_value < alpha:
        print("✅ Se rechaza H0: Existe relación monótona")
    else:
        print("❌ No se rechaza H0: No hay evidencia de relación")
    
    # Interpretación de magnitud
    print("\n Interpretación :")
    if abs(tau) < 0.2:
        print("Relación muy débil o inexistente")
    elif abs(tau) < 0.4:
        print("Relación débil")
    elif abs(tau) < 0.6:
        print("Relación moderada")
    elif abs(tau) < 0.8:
        print("Relación fuerte")
    else:
        print("Relación muy fuerte")
```

```python
# Ejecución de la Prueba de Kendall #
prueba_kendall(x, y, alpha=0.05)
```

En muchos problemas reales, los datos no son números continuos sino categorías: género, tipo de producto, respuesta sí/no, nivel educativo, entre otros. A simple vista, comparar proporciones puede parecer suficiente. Sin embargo: las diferencias observadas en los datos pueden deberse al azar y no a una relación real.

Por eso usamos pruebas de hipótesis para variables categóricas, que nos permiten:

- Determinar si existe dependencia o independencia entre variables

- Validar si una diferencia en proporciones es estadísticamente significativa

- Tomar decisiones basadas en evidencia, no en intuición

En este contexto, herramientas como la prueba Chi-cuadrado se vuelven fundamentales para responder preguntas del tipo:

## **Prueba de Independencia Chi-cuadrado**

### **Objetivo**

Evaluar si dos variables categóricas están relacionadas o si son independientes.

---

### **Planteamiento de hipótesis**

$H_0$ : Las variables son independientes (no hay relación)

$H_a$ : Las variables son dependientes (sí hay relación)

---

### **Idea clave**

Se compara:

🔹 Frecuencias observadas (datos reales)

🔹 Frecuencias esperadas (si no hubiera relación)

👉 Si hay mucha diferencia → hay evidencia de dependencia

---

### **Estadístico de prueba**

$$χ^2 = \sum\frac{(O_{ij}-E_{ij})^2}{E_{ij}}$$

🔹 $O_{ij}$ : Frecuencia Observada

🔹 $E_{ij}$ : Frecuencia Esperada

---

### **¿Cómo se calculan las esperadas?**

Bajo independencia:

$$E_{ij} = \frac{(total\ fila\ i)×(total\ columna\ j)​}{total\ general} $$

</div>

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">  


### **Intuición de la Prueba**

“Si las variables fueran independientes, los datos deberían verse *Balanceados*"

"Si no lo están → hay relación”

 </div>

```python
import pandas as pd
from scipy.stats import chi2_contingency

def prueba_independencia(df, var1, var2, alpha=0.05):
    
    print(f"\n{'='*55}")
    print(f"🔍 {var1} vs {var2}")
    print(f"{'='*55}\n")
    
    # =========================
    # Tabla de contingencia
    # =========================
    tabla = pd.crosstab(df[var1], df[var2], margins=True)
    print("📊 Tabla de contingencia:")
    display(tabla)
    
    # =========================
    # Chi-cuadrado
    # =========================
    tabla_sin_totales = pd.crosstab(df[var1], df[var2])
    
    chi2, p, dof, expected = chi2_contingency(tabla_sin_totales)
    
    print("\n📈 Resultados prueba Chi-cuadrado:")
    print(f"Chi2: {chi2:.4f}")
    print(f"p-value: {p:.6f}")
    print(f"Grados de libertad: {dof}")
    
    # =========================
    # Decisión
    # =========================
    print("\n🧠 Decisión:")
    if p < 0.05:
        print("✅ Se rechaza H0 → Las variables SON dependientes")
    else:
        print("❌ No se rechaza H0 → No hay evidencia de dependencia")
    
    # =========================
    # Frecuencias esperadas
    # =========================
    expected_df = pd.DataFrame(
        expected,
        index=tabla_sin_totales.index,
        columns=tabla_sin_totales.columns
    )
    
    print("\n📊 Frecuencias esperadas:")
    display(expected_df.round(2))
```

```python
# Función de Pruebas que Incluye mapas de calor #
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import chi2_contingency

def prueba_independencia_pro(df, var1, var2, alpha=0.05):
    
    print(f"\n{'='*60}")
    print(f"🔍 {var1} vs {var2}")
    print(f"{'='*60}\n")
    
    # =========================
    # Tabla de contingencia
    # =========================
    tabla = pd.crosstab(df[var1], df[var2])
    print("📊 Tabla de contingencia:")
    display(tabla)
    
    # =========================
    # Chi-cuadrado
    # =========================
    chi2, p, dof, expected = chi2_contingency(tabla)
    
    print("\n📈 Chi-cuadrado:")
    print(f"Chi2: {chi2:.4f}")
    print(f"p-value: {p:.6f}")
    print(f"Grados de libertad: {dof}")
    
    # =========================
    # Decisión
    # =========================
    print("\n🧠 Decisión:")
    if p < alpha:
        print("✅ Se rechaza H0 → Hay dependencia")
    else:
        print("❌ No se rechaza H0 → No hay evidencia de dependencia")
    
    # =========================
    # Frecuencias esperadas
    # =========================
    expected_df = pd.DataFrame(
        expected,
        index=tabla.index,
        columns=tabla.columns
    )
    
    print("\n📊 Frecuencias esperadas:")
    display(expected_df.round(2))
    
    # =========================
    # Validación (regla de 5)
    # =========================
    bajos = (expected < 5).sum()
    total_celdas = expected.size
    
    print("\n⚠️ Validación supuestos:")
    print(f"Celdas con frecuencia esperada < 5: {bajos}/{total_celdas}")
    
    if bajos / total_celdas > 0.2:
        print("⚠️ Advertencia: Muchas celdas con valores bajos (resultado poco confiable)")
    else:
        print("✅ Supuestos razonablemente cumplidos")
    
    # =========================
    # V de Cramer
    # =========================
    n = tabla.values.sum()
    k = min(tabla.shape) - 1
    
    cramer_v = np.sqrt(chi2 / (n * k))
    
    print("\n📏 V de Cramer (fuerza de asociación):")
    print(f"V: {cramer_v:.4f}")
    
    # Interpretación
    if cramer_v < 0.1:
        print("Relación muy débil")
    elif cramer_v < 0.3:
        print("Relación débil")
    elif cramer_v < 0.5:
        print("Relación moderada")
    else:
        print("Relación fuerte")
    
    # =========================
    # Heatmap (🔥 visual potente)
    # =========================
    plt.figure(figsize=(8, 5))
    sns.heatmap(tabla, annot=True, fmt="d", cmap="Blues")
    plt.title(f"Heatmap: {var1} vs {var2}")
    plt.xlabel(var2)
    plt.ylabel(var1)
    plt.show()
```

## **Mutual Information (Información Mutua)**

### **Definición**

La Información Mutua (MI) mide cuánta información comparte una variable con otra.

En otras palabras : ¿Cuánto saber $X$ reduce la incertidumbre sobre $Y$?

---

</div>


<div style="text-align: justify; line-height: 1.5; font-size: 18px;">  

### **Definición Formal**

$$I(X;Y) = \sum_{x \in \mathcal{X}} \sum_{y \in \mathcal{Y}} 
p(x,y) \log \left( \frac{p(x,y)}{p(x)\,p(y)} \right)
$$

https://link.springer.com/chapter/10.1007/978-3-642-04138-9_30

---

### **Interpretación intuitiva**

🔹 $MI = 0$ → Variables independientes

🔹 $MI > 0$ → Hay dependencia

🔹 Mayor $MI$ → Mayor relación

---

### **A diferencia de Chi-cuadrado:**

🔹 No depende de supuestos fuertes

🔹 Detecta relaciones no lineales y complejas

</div>

```python
import pandas as pd
from sklearn.feature_selection import mutual_info_classif

# Variables
cols = [
    'situacion_laboral_del_jefe_del_hogar',
    'nivel_educativo_del_jefe_del_hogar', 
    'genero_del_jefe_del_hogar', 
    'tenencia_de_la_vivienda', 
    'zona_urbana_rural'
]

# Copia
df_encoded = data[cols].copy()

# Codificar categóricas (clave)
for col in df_encoded.columns:
    df_encoded[col] = df_encoded[col].astype('category').cat.codes

# Definir X e y
X = df_encoded.drop(columns=['tenencia_de_la_vivienda'])
y = df_encoded['tenencia_de_la_vivienda']

# Calcular MI
mi = mutual_info_classif(X, y, discrete_features=True)

# Resultado ordenado
resultado = pd.Series(mi, index=X.columns).sort_values(ascending=False)

print("Información Mutua con Tenencia de la Vivienda")
print()
print(resultado)
```

```python
# Graficando la Información Mutua #
import matplotlib.pyplot as plt

resultado.sort_values().plot(kind='barh')
plt.title("Importancia de Variables (Mutual Information)")
plt.xlabel("MI")
plt.show()
```

## **3. Variables Continuas vs Categóricas**

### **ANOVA**

### **Definición**

Evalúa si la media de una variable continua cambia entre grupos definidos por una variable categórica.

¿Los grupos son realmente diferentes o las diferencias son puro ruido?

---

### **t-test**

### **Hipótesis**

- $H_0:$ Todas las medias son iguales
- $H_a:$ Al menos una media es diferente

---

### **Idea**

$$F = \frac{\text{Varianza entre grupos}}{\text{Varianza dentro de grupos}}$$

---

### **Interpretación**

- F grande → diferencias significativas

- Si los grupos son muy distintos → F grande → ❌ Rechazamos H₀

- Si son similares → F pequeño → ✅ No rechazamos

</div>

## **ANOVA No Paramétrica**

### **Prueba de Kruskal–Wallis**

Es una prueba no paramétrica que evalúa si las distribuciones de una variable continua difieren entre varios grupos (definidos por una variable categórica). En lugar de comparar medias, compara rangos, ordena todos los datos y analiza si los grupos tienden a tener valores más altos o más bajos.

Es el “ANOVA sin supuestos fuertes”.

---

### **Hipótesis**

$H_0:$ Las distribuciones (medianas) son iguales

$H_a:$ Al menos una es diferente

---

### **Estadístico**

$$H = \frac{12}{N(N+1)} \sum_{i=1}^{k} n_i \bar{R}_i^{\,2} - 3(N+1)$$

---

### **¿Cuándo usarla?**

- No hay normalidad
- Hay outliers
- Variables ordinales o continuas no normales


| ANOVA               | Kruskal-Wallis |
| ------------------- | -------------- |
| Usa medias          | Usa rangos     |
| Paramétrica         | No paramétrica |
| Sensible a outliers | Robusta        |


 </div>

```python
# Función Base para la Prueba #
from scipy import stats

def kruskal_test(data, x, y, alpha=0.05):
    """
    x: variable categórica
    y: variable continua
    """
    grupos = [
        data[data[x] == cat][y].dropna()
        for cat in data[x].dropna().unique()
    ]
    
    H, p = stats.kruskal(*grupos)
    
    print(f"📊 Variable categórica: {x}")
    print(f"📈 Variable continua: {y}")
    print(f"H = {H:.4f}")
    print(f"p-value = {p:.4f}")
    
    if p < alpha:
        print("❌ Se rechaza H₀ → Hay diferencias significativas entre grupos\n")
    else:
        print("✅ No se rechaza H₀ → No hay evidencia de diferencias entre grupos\n")
```

```python
kruskal_test(
    data,
    x='nivel_educativo_del_jefe_del_hogar',
    y='ingreso_del_hogar'
)
```

```python
kruskal_test(
    data,
    x='zona_urbana_rural',
    y='valor_de_mercado_de_la_vivienda'
)
```

```python
kruskal_test(
    data,
    x='acceso_a_credito',
    y='indice_de_pobreza'
)
```

## **Función Unificadora de Pruebas**

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from scipy import stats

#=====================================#
# FUNCIÓN ANÁLISIS COMPLETO
#=====================================#

def analisis_completo(data, x, y, alpha=0.05):

    #=====================================#
    # LIMPIEZA
    #=====================================#

    df = data[[x,y]].dropna()

    #=====================================#
    # BOXPLOT INTERACTIVO
    #=====================================#

    fig = px.box(
        df,
        x=x,
        y=y,
        color=x,
        points='outliers'
    )

    #=====================================#
    # PUNTOS DISPERSIÓN
    #=====================================#

    fig.add_trace(

        go.Scatter(
            x=df[x],
            y=df[y],
            mode='markers',
            marker=dict(
                size=5,
                color='black',
                opacity=0.30
            ),
            name='Observaciones',
            hovertemplate=
            f'<b>{x}</b>: %{{x}}<br>' +
            f'<b>{y}</b>: %{{y}}<extra></extra>'
        )

    )

    #=====================================#
    # CONFIGURACIÓN
    #=====================================#

    fig.update_traces(
        selector=dict(type='box'),
        boxmean='sd'
    )

    #=====================================#
    # ESTADÍSTICAS DESCRIPTIVAS
    #=====================================#

    resumen = (
        df
        .groupby(x)[y]
        .agg([
            'count',
            'mean',
            'median',
            'std',
            'min',
            'max'
        ])
        .round(2)
    )

    print("\nESTADÍSTICAS DESCRIPTIVAS\n")
    print(resumen)

    #=====================================#
    # AGRUPACIÓN
    #=====================================#

    grupos = [
        df[df[x]==cat][y]
        for cat in df[x].unique()
    ]

    #=====================================#
    # ANOVA
    #=====================================#

    F,p_anova = stats.f_oneway(*grupos)

    #=====================================#
    # KRUSKAL
    #=====================================#

    H,p_kruskal = stats.kruskal(*grupos)

    #=====================================#
    # RESULTADOS
    #=====================================#

    print("\nRESULTADOS ESTADÍSTICOS\n")

    print(f"ANOVA → F = {F:.4f} | p-valor = {p_anova:.6f}")

    if p_anova < alpha:
        print("❌ ANOVA: Hay diferencias significativas entre medias")
    else:
        print("✅ ANOVA: No hay diferencias significativas")

    print("\n----------------------------------------\n")

    print(f"Kruskal-Wallis → H = {H:.4f} | p-valor = {p_kruskal:.6f}")

    if p_kruskal < alpha:
        print("❌ Kruskal: Hay diferencias entre distribuciones")
    else:
        print("✅ Kruskal: No hay diferencias significativas")

    #=====================================#
    # CONCLUSIÓN AUTOMÁTICA
    #=====================================#

    print("\nCONCLUSIÓN FINAL\n")

    if p_anova < alpha and p_kruskal < alpha:
        conclusion = (
            "Ambas pruebas coinciden: "
            "existen diferencias estadísticamente significativas entre grupos."
        )

    elif p_anova < alpha and p_kruskal >= alpha:
        conclusion = (
            "ANOVA detecta diferencias, pero Kruskal no. "
            "Posible efecto de outliers o incumplimiento de supuestos."
        )

    elif p_anova >= alpha and p_kruskal < alpha:
        conclusion = (
            "Kruskal detecta diferencias, pero ANOVA no. "
            "Posibles distribuciones no normales."
        )

    else:
        conclusion = (
            "Ambas pruebas coinciden: "
            "no existe evidencia suficiente de diferencias entre grupos."
        )

    print(conclusion)

    #=====================================#
    # LAYOUT
    #=====================================#

    fig.update_layout(
        template='plotly_white',
        title={
            'text':
            f'{y} vs {x}'
            f'<br><sup>ANOVA p = {p_anova:.6f} | Kruskal p = {p_kruskal:.6f}</sup>',
            'x':0.5,
            'font':{'size':28}
        },
        xaxis_title=x,
        yaxis_title=y,
        width=1300,
        height=750,
        showlegend=False
    )

    fig.update_xaxes(
        tickangle=20
    )

    fig.show()
```

```python
analisis_completo(
    data,
    x='nivel_educativo_del_jefe_del_hogar',
    y='ingreso_del_hogar'
)
```
